# Exploring our original news datasets and filtering them to only what we need

```python
import pandas as pd
import warnings

warnings.filterwarnings(
    "ignore", category=FutureWarning, message=".*un-recognized timezone.*"
)
```

```python
df_cnbc = pd.read_csv("../../data/news/original/cnbc_headlines.csv")
df_guardian = pd.read_csv("../../data/news/original/guardian_headlines.csv")
df_reuters = pd.read_csv("../../data/news/original/reuters_headlines.csv")
```

### I'll start with just the CNBC data and then do the other two at the end once I know what to do

```python
df_cnbc.head()
```

It looks like we have some null rows, lets drop those

```python
df_cnbc.info()
```

```python
df_cnbc.dropna(inplace=True)
df_cnbc.reset_index(drop=True, inplace=True)
```

```python
df_cnbc.head()
```

```python
df_cnbc.info()
```

Around 280 rows had empty entries for some reason, hopefully after adding the data from other news sources we will still have enough headlines to work with. We still have 2800.

Let's also check the range of our data as we need to know this for later when we start matching news to stock prices.

```python
df_cnbc['date'] = pd.to_datetime(df_cnbc['Time'], format='mixed')
min_date_cnbc = df_cnbc['date'].min()
max_date_cnbc = df_cnbc['date'].max()
print(min_date_cnbc)
print(max_date_cnbc)
```

To make things easier I'll combine the headline and description into a single summary field.

```python
df_cnbc["summary"] = (
    df_cnbc["Headlines"].fillna("").astype(str).str.strip()
    + " "
    + df_cnbc["Description"].fillna("").astype(str).str.strip()
).str.strip()

df_cnbc.head()
```

I'll save the cleaned data to a new CSV file for later use when we make our models. Before I save I'll also specify which source it came from in case we want to know that later and maybe compare how accurate different sources are?

```python
df_cnbc["source"] = "cnbc"
```

```python
df_cnbc[["date", "summary", "source"]].to_csv("../../data/news/filtered/cnbc_filtered.csv", index=False)
```

We will fully clean the data in a separate notebook later, for now we just want to get rid of any obviously unusable data.

I'll do the same for the other two datasets now.

## Reuters data filtering

```python
df_reuters.head()
```

```python
df_reuters.dropna(inplace=True)
df_reuters.reset_index(drop=True, inplace=True)

df_reuters["date"] = pd.to_datetime(df_reuters["Time"], format="mixed")
min_date_reuters = df_reuters["date"].min()
max_date_reuters = df_reuters["date"].max()

df_reuters["summary"] = (
    df_reuters["Headlines"].fillna("").astype(str).str.strip()
    + " "
    + df_reuters["Description"].fillna("").astype(str).str.strip()
).str.strip()

df_reuters["source"] = "reuters"

df_reuters[["date", "summary", "source"]].to_csv(
    "../../data/news/filtered/reuters_filtered.csv", index=False
)
```

Worth noting that Reuters data only has the day and not the time, so we may need to account for that later when matching to stock prices.

## Guardian data filtering

```python
df_guardian.head()
```

Already only has one column for the article text so we don't need to combine anything here.

```python
df_guardian.dropna(inplace=True)
df_guardian.reset_index(drop=True, inplace=True)

df_guardian["date"] = pd.to_datetime(df_guardian["Time"], format="mixed", errors="coerce")
df_guardian.dropna(inplace=True) # remove rows where date conversion failed
df_guardian.reset_index(drop=True, inplace=True)

min_date_guardian = df_guardian["date"].min()
max_date_guardian = df_guardian["date"].max()

df_guardian["summary"] = (
    df_guardian["Headlines"].fillna("").astype(str).str.strip()
).str.strip()

df_guardian["source"] = "guardian"

df_guardian[["date", "summary", "source"]].to_csv(
    "../../data/news/filtered/guardian_filtered.csv", index=False
)
```

### Now that we have separate CSV files, lets see what ranges each used

```python
print("CNBC date range:", min_date_cnbc, "to", max_date_cnbc)
print("Reuters date range:", min_date_reuters, "to", max_date_reuters)
print("Guardian date range:", min_date_guardian, "to", max_date_guardian)
```

It looks like if we combine them into one dataset, we will have data from 2017-12-17 to 2020-07-18, although the first three months won't have data from all three sources. That won't be an issue unless we compare sources, but that isn't the plan right now so let's just combine it all.

```python
df_cnbc_filtered = pd.read_csv(
    "../../data/news/filtered/cnbc_filtered.csv", parse_dates=["date"]
)
df_guardian_filtered = pd.read_csv(
    "../../data/news/filtered/guardian_filtered.csv", parse_dates=["date"]
)
df_reuters_filtered = pd.read_csv(
    "../../data/news/filtered/reuters_filtered.csv", parse_dates=["date"]
)

df_all = pd.concat(
    [df_cnbc_filtered, df_guardian_filtered, df_reuters_filtered], ignore_index=True
)
df_all.to_csv("../../data/news/filtered/all_filtered.csv", index=False)

df_all.info()
```
